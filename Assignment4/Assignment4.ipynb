Build a simple question-answering chatbot using a pre-trained language model.

```python
from transformers import pipeline
```

```python
qa_pipeline = pipeline(
    "question-answering",
    model="distilbert-base-uncased-distilled-squad"
)
```

```python
context = """
Artificial Intelligence (AI) is a branch of computer science that focuses on creating intelligent machines.
Machine learning is a subset of AI that allows computers to learn from data.
Deep learning is a technique within machine learning that uses neural networks with many layers.
AI is used in many applications such as robotics, healthcare, self-driving cars, and recommendation systems.
"""
```

```python
result = qa_pipeline(
    question="What is Artificial Intelligence?",
    context=context
)

print("Answer:", result['answer'])
```

```python
def ask_question(question):

    result = qa_pipeline(
        question=question,
        context=context
    )

    return result['answer']
```

```python
print("Question Answering Chatbot")
print("Type 'exit' to stop\n")

while True:

    question = input("Ask a question: ")

    if question.lower() == "exit":
        print("Chatbot stopped.")
        break

    answer = ask_question(question)

    print("Answer:", answer)
    print()
```

# Explanation

Question Answering Chatbot using Transformers

This project implements a simple question-answering chatbot using a pre-trained transformer model. The DistilBERT model fine-tuned on the SQuAD dataset is used to extract answers from a given context.

The system works in the following way:

1. A context paragraph containing information is provided.
2. The user asks a question related to the context.
3. The transformer model processes the question and context.
4. The model extracts the most relevant answer from the context.

Advantages:
- No training required
- Uses powerful pre-trained transformer models
- Provides accurate answers from text

Applications:
- Educational assistants
- Customer support chatbots
- Information retrieval systems
- Intelligent search engines
